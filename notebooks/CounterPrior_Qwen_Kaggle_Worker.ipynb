```python
PAYLOAD = {'avh_common.py': '', 'avh_kaggle.py': '', 'avh_kaggle_worker.py': 'H4sIAAAAAAACCsU8aXPjRnbf+St64Q8BbQojTTyuFB26SjuW7YntGa9nvFsJw+qCgCYJEwRgNCiNVlF+e97VQAMEJU3tVmWqbOHo491XPzAIgu9qY7dnVV0mxlr1/S+/qduy3pk6Uq9/+/ZS//XN+zd//ulKf3v11zevr96rzCprGnV9p5qtUUXZmOuy3Kkqrk3RRJPJ21LFSVIeikbl5SYrVHxoyn3cZGURKXjZGNso28AD22SJjdS7wqh9mZpcZQU8LxKjygLmJnGukkMaz8+jSRAEk3Vd7pXW60NzqI3WKttXZd2ouAAYaHk7mbhn9QbgscbdJ2Wem4THyKNN4q62sd3m2bW7/d2WhbsuLe9axQ0OcVv+ArduiN0emix3d022Nx0UN1udlPs9YBNbvPNf7OLNJjf4YreZTCapWavNIa5TbfZVc6dtWsX6j4OpM2PDfWx3eh0nTVnfzRSQO9dpc1eZ6Xyi4B8Q50djKrU+5PmdwsEmBZDTNCs2StZQ66zIGqNsiWy7UwmRDQhTNPE+K2J4VRsgeFPuTGEjpDcu3YJl0vBzoKqdqc8/393ilWyO/3BLtVA+mEej28HZmseDHCEEb0tgf1yk9DAitNRi4SHZbYP/agPcL3iF/1H/S5Pi4i5Ms/3i7GKmdkAIvP5QH0y3pzdr4t0LZkD+JI9B9v8cN8n2fVLWpm4pS9pBpFXrsgZcUqtus2argGbZjVGiN2XtKP61KsyNgdvarLOPZ0mcbJG2B2t6VNUaGaJ1aE2+nilLu3o0xccRPwXS8kX/JZA52bbv+LYbwQ9b6CJibPZ3fMZwapulQGoV5GbdBJMWMJopUNXl7QwQ3RvgZFak5uNMpVm8KUpU3cV3cW6NB7JId3HYV3co2EXVviLgAEsEt4V95iPZHxqh4kf2rki2dYlghx0zYWuUNlS1CHSrbGBA4r83HxuEdx9vjPurN3V5qOAuPqRZiWAsVzPvv3Yy8JgQBnQZ8b74ZXu3xkxpRCbaG6BI2BJK6DTtzUItA+MI48PlfYBCHcxVQIAFD7Slxv1g8ZX6QnlDaKfgYTVFrYHXygDFfXD91b9Y+FORCsFM/sL9fxd/OYDlBRMIN7ALQLwM/pBHAe4LQy4Lewvy9p9gLwAoMNZlkd9FAEBvQyJwFFeVKdLQejIGj/I7nWzjRjdgxnKwKohwXeYEEqhAjSAJwPBIrh5Ws94GLAYsryxlQPQ01RtTmJpMvYZNwU6Sms+UKeLr3OhmmxU7kGwRzD4PPlPfm/0+VjdxnqUAGGAIhod4oEg4VAWYI2os9VmR5AeyoGSQeYwdLEnSUp6RDOA08GaXag14y8I5+Di1N3Fh1bu3V8rGQBSAF+3en+XORgMB66TVkRhXCYH7QynoI8jv5keEZD2IADFcC5c5GsJK4baju24QWm+Q3HsnSaJegZgReMJMCNioAucLkAWLDK+a4GHSgw9B6YOIy0eHClkS8vuFTwOcZqNkvYk4RADXEGyQkQFTQbT8CKUeagtGcMYkR9sHUmQWF1+dn5/3HNNwpz9uTREcw7tEUda0KFhxfQOGtAQVWigSvHa4KRJYKSUz0SpJ+PnnA39YVIeGKLybqxuwjGFqbrLELAK0gaAv5AUXbBbXeRk3F18hrDdRZjXdI0JVmRVNOGWi3LAfnY5QBSV2N1M3aG8Evgjigr0NWbqsi8HCm5mY4g/E0OnDJxIqN8Wm2SJijGG0MU0YrE1M4VvcoOKjJqNTDqYnpspVZA/78OyCQHQv/fiBsMar3jJVaTOK+VpbjfAiHLouK6PJUocM3TKgvzpLbbCa9UB24pilYGFug+mxsOHObs4yGKAGq4kJE8hnNH6gvDLXgcxwANAtDkNdh6gVNqpDwQr1D3GCyya2wfTYDjj0vWGwfo9oSMb58+AKT2Eb5WWxCafgwPeOaWfqAu5z0D0NoWboqVxhTKq3WZqaAtbsQgv0PgHGXEHrh9s5FHyxZGbF2kDWkRiNqIUDpMtDAyB2jAfNY6BnCtWXAjPnXcR2pVnifArPFuA05itgmTx4OyQwyyEF5ikR3y/nM3V2sWKNDXtWBsVWELAcloc8ZxrFeT5Eo44zEO7vRNF/QT2/qmswJOsAmCdh/boCoyCArCkWtRZc3D2R7uFrF5su7u2TIeGDp4wtaxhcoEKyC5fbAW604xY5JRToEW2FMeN+wcrrM/xYaVGs+d0wOQhHaCZ8YJop8sQnxkTbOF+HUxn6XPp25H1RQkAPyN6CMSdKC5SMoEevO2NbYgE3QAHMxyrsBMIyzUFtYGSwWiEZPW0onzm5KI/mQpZyyBuKa3uxbDZrw1kDgTmGTyYkqRiQYQ9eKUNWM1MBvmW2AtUtSvg7HeRguJcLF+4DFDZw9xz+BrwQ3PMFPIkpqoQn6KJkn2/U+YghRbLoFneY4QMzhaWKcuQ1gziyWmXinb6p472+vkNGzf3kYh9/1Huzh1xVg1iUCRAmDcdWYWbDZL6AvaKkOoCJo0SH/eZJsUXao4NkYwYM6YzctNOAkV2Z77CrCMDYrp7BbJd66HiFzpk10jmnmfP5M1l2JpDPUHQxMH1+GmbyuLIU3QwzMZAazNGG2TeLzcxNlKIHGOFkp6s4q0NImNmcz7A8kVKYLlJ6IrXM01Zgiwo5in/ia9sttXSERFluV+2eTiVN2D5jJRGEwUruqVsJQ5QU5Ds6f4n83nrX3lqiEyv1p4W/mHvc6Sabp18PBVLZGX4qVLxYU2miiusM8pN1nOUmdeKyuM/TB8fcxf02dVZdeHEvBNCoBYAmeL71GoUNJEMwGr6CNYRlLmvU1wiGsSH9nbm8icoEFoRFSgcLDo9OFA5oCEZ6mOTwDYaom4hdmAYjhtn11IHPMgTEAGPd5fETZ+/AJIOcgIGp42JjwnOKusTgMVSe3atw7IJBWNLMOc//gkZ2lpS1CBMIAylrivsSzly74OwfFxurkngKP7DWLieT1f0KB2HXt7Bkv3iTQMCAB3IFzwggjXAjHwFpHDt96LF9QD7hJ+F8qLTdg39xzEwgT77mZLvFiuvCEAPOkNYQdkFCn8V5j59BELxGleaSmcQhZQFpY9EKjiXjCFYXAPrjkNVUrnQD19lNVymzkoT0CV7P5K6reTEkqMMeYChJ48kjayrmPRgfgmkB29/DeGfuFhCxXqeQ189VWC+DhsJbXCO+gaTo8v2P75cvV6OZJw5IM8hLm3CJ1++vrr4FDjH5zojMQY+asHyWgk8HG7KcX5yvWkMCbCQop2gnLs4fNQvBWwiuYJDPOsERNQOr9bylYgicZICXbGadbXJKdf8w8YIIEPF+DMFgdQCts9o2nGgdqcaS3d9qTD36hVoW5394GUbK6Y/nYgjK5fnKaTJcTnuxeab+vUdn/CdlJ8nkvSIULRx2Zu+oINPOXAbtpfZNv4NnzPI/xurvZbUXSB4sUqWZjTe1MXssrB1h9GpYRF9TweFhMiwNuFjx2P6OwAWrLHHCapRZ9OYkqxDn89UQKpHAJRUnSSNgadyHBrZ+EmuwSWIqZsn9Q98DgAVEFMLwYkZRktgCipDYK1zM1Fd+GtDUd3300OXZfjDtlhdFIPFfzl+N8Kt1FijEz/eVJ4W5B5TzGWNhk0+3JZgUissxUag5Rau7FG1QBOwoC26wDslNrqjohynOgYt5KA8p12/3GHRYMOTiagi0sdgZYwiJNCSMwNCqXo6GH6sOykcXlPikv+BY0HK0YF/gnQy1pWxEu4tvP+JbPw5+d2jerX+mlIHUcD5Ud5a0ExrcNylJJGeSYZ8T3nZ8EEhlivDT2QV81uVaMhwpw0rtweH9hDcpwTo1mBNLiHkGSXC2zkDnSIZJbIdRpQhJWxMudOvz4eEFhGOBZz9Jifh5C0rQinP7+tXMf7oDrUEEKfrVh4IVK9Wb6hB4y/DxG5NPO8hhnhR8PCF2K4Y9GgfMTI0uR8vJn2a/El9nOdAjGBqX4PRYMPopmwDNxPTLiP3QDeXZcQhTXndg4ebjMKSKG9MjnQ/evPXtQScweOjS3jx0ASCen5jiJoM0D11I2DuPhGjshxhSDXL+GZ7L1ywqVDiLQb0wuECK5gYPkuNbEJlk156AxNgyAHjjKXNVlzemwApzG+RJcsd/IHSJ9rAoABj3EoQwIN2gMy0w4xboDNBYvKe0EC9+ySCTv8Wr+Ab/b8tDka4hMcIbDIpKG9NboF1ODjTw8wopwsvZKZBjuWMTsiOHElR3zbYsCIJToJAUKle696qj10AQMDKB2CDeEG000YmLqlSum48QIgIBtrAKJTLT40NKf013mCKRAZVGFz3RBMCwigUSmh4SkCtQlSTD5fkJgrsYCWv9c+ZI0LGultLsDzmW18pb/dTaIxmJyCGHp+Hv5bUI3mfqjfRVUBKAp5FqC3YA/P+6weO6bQbBLafB0r+yBWHDKij2sXgCHfmS1qUNoMGljWQcl/vHml4Cir/R1gJs6LxAB8jgrKaPG9G/ccCNwIBlzDYEi9pndo+6HLQVA8/s8+mPpg6akON+Kod/+DIg643a1o3GwwyZgZIQnj8B0NXHOGmAilihugHGgN6rD1/S2ackBy4l64AjnFsDzIHrqPl9fOtf3QoqFZlQ6HbxDJG3Qkm3pqHaO2gDdtkwveVFgCXA7ilHMoEUJ+Eu2u9SiIe4E8m6s+GPYIF0ufOCqmswCjm2PmBuVps41djzE8ou6oUKuDlH88AI3wqIrQscTiaIkurQuZpVSz/3aOncM1OwNxpZ3I3jjbXdxi9ffcXDvZyS3z7BaLgjq/waRE8Q5vUTTgwp8whGYKTDbD4bOwXAU3TzlhDiPQWsoQIseiHFh+k0WSVbjNdT0XZExcdBoJfcnsWUboI2gR6cUHZnCTCY1Z3qAdr5roAOEtqXOImOKiEUYxtw8QQmf4EZ6B0tG6tSutksCqVomBV0uB/B7eyOEfZxka2xRW5UwtxbT7rcoxZ3BJPxBlS6t6wW2j3o8ZYF5PSYJ4wc2M6NtDO8sM0hxWrkPsvvBiYEgYL9yJUCicg/pinysVt/gzHCouMAhOYQMMR5gAWJXl4b4lAeJGpFDJIZolLdmL5KTcdV6jiX85cotDCQpxNO3bPV6LlSn1Lvbgv0WlR9o77IM8IXnBi4BIsPkWXiHVoTjDvw8SLqGBFYG2w6QGvcjhDrN2LvJJIF91Cke+5DIiuqNYZGWk8jPFNl/xF8f/Xzz5f6/dWvby5/0pc/X739Fv77EO09fzAspnlNa9QMCalNdYdLh8OduegFWAxfRBTYTIZlCIj5TL23EDvcZOaWbQklA7kBkwMWBkJJ7DZs37A2/wQ+++37q6jhFijSYv/Z9DTEnTHj4yyBtwMvlCdEhOlpovfRoNG6DeccuC5G6+jA5vI58DEXBnD6DxkOn+un7LabdRKeyfNAaactXfrlxJQkDSt3AzifnCFIAOBc8SB5IB7MhAVsNYGyz0FwjA+8QrLeyAKv8dh3E2K0h/tMZaMZbYKB0A0FL4jNbvHS1XHtYmgPusTMVXrRp/GwtvaLqzwSqXfoQxxdQlZ5xKEQABfNJJe5wNodxGZlDjZQWlyEEMJGrPaPnttJC2ic5lmBRtiN/EKFF/92fj7mTynvvTiPXqnP1b9+5bqp/AL0Qi3rrgzTuSPslkNnsMbit60gY+XieuBNFj9OHR9Ig7+hWUzLTYhbzVpQF+5iipkmvJ/3mo1PWivOMD5gDlGbDWhwDQ4bAowaG2Pz7MaIE5f8NoXAhE6JxXNLl3y/Qojd4n5OGGFXDDIMdce6BOTyp5/0z5fvf9SXHz6AdX3z7q3+7re3r/Hi/eBU3u4w+f20CUDQtIoppXy0p7wd6Fq9sOt6OughFq14ty8ybo3uJI67X0DdumOxkbNgEaOJfzoSN+U+S1hNxa/V8a1fgXA2skvGuz25/rTw+7VdwaI7/C43kKoekq1XO6NzD81HfI80ENMwJ//PORc7Vune8Zgf/DjVwfJQ3o9jPQgfJWeHxLiYd5HuaWaobxSq9NMRC8WzkCyeSRRM/H7BFKLYBcukBlTja+VhNEjsHuE6rdTyGm/6LBw1R57B6Q3u6DZ8M5QE9DENHptT0RRcCxoGtJzvX/9w9fMlHqf24sX5SLA4G9YL5RMTLcHvnIpJrcUblJLGI20MYbDvVXfVAu+VNVyA3wXTh8HuHPvPnasSR9f6qjn6t8jdUmHW+Sl+JXeDVbFXRG+yRuM3MjCyDcv7L2ZtAtF/Ti079AiCJKpI9hfgpzMqFk/Hth6jvhTJ8AmFr7hTOIhnmbq9OOgZ//ohW/f1TsTFxe6rnaMHQkx8Pn0YYuIZNTywP1lr9avcTKpe1WWwaluL1TmkE7hydP7SCynJ3szVoPA3EFl0COwb0C3cafIWToo8n/nwfIfqNGsZDGP8LuNcDLg3HDk9seKwZo8rkc0YeXViCa/U7s32n/qoHqWtnLoR8yF7G7bZfqauig0YK4McOMPYYa5enasSs74z91Fc1wRBUY96+erFy1cUd1DdsrktsVw4/CxAWhXAQkGYAjYWazW6icEv9RzSKwiLXMfBdHQFPrgcljHn85eDbzy8BptHz0AfOf/EBhT73DPQY0Yc6+1tDVs+7b87dvyKn7RRxyaLWPZ36YqwNBucFlWQvT46/nYLpAXrNA19uWVNjMXn38tDXcR5dLQNNbosYObHJqyPDlolwuWz2K4VcDryncbJr6faBtEqsvGNcb6TeSTHoVFBRgkX73rSotjihwDYyyZfDkxHqIrh9WBNPFi25JYB1KgEdw4GMaD4ej2fnDKePt7zk1Z3HREfQ1w9SgFdG9KnD71vE2r/q4SdK3yHDrNZ2yAJ9pa/XQrG7fwaMD/YbTj+trTRGtsbwzWZo6IMpyfF7rGgzJPM4y+Zht1crT5kaGvbM3s+r8K4kUIA96r3bLzfSHFzkkxo5KuDQNRN2sawR4xyaPpURHctZD30Tqzfgux1nh32YbVsW9E4uqmQS7QbcqW38sNIR8NIZPZIrMiEdMHifeBOgk1ndLXIL/BGO83VfMZkg7F2gkHY1AXx85GoPnB+BHuZ5XKsdTc+FLBtt0j3YEWC22VM1N/b3c4kIPbJ7N+PbNY7Ox6xovRhyC31PrMcDvhw2phMo0MBofculNpkv7414iHRuJFb7Ov+bQzx4LOM9u0WdJB0vTvlwSTD1YIiqrPZcKQZCc/QRtSy2/obzAnGjdJY3vNBarSbTW027DTSLCXYqrrEs03IsLwSLhpREP2LcwXzDv2vAjxDYCKbG1OFF/23UvUeKfj76B91l1HLcO9MqS1+P6+N7MOwEr2N8wbdYlbYCrIZzOnQKV0bUG7E+A4i/XyHBxwD/EaTrL54IHwSPx11zjyZIo9ynCPRP7U5b++8o+AtL87Pz/GFl+Tg4y+fLwp8vMEHO16Ka00DBnnEj4thcOfwnTIuLmZqEKcuToWvx+tWNX4+4TvNnrXq8NMSUgVYgoIAjy1IvzPnGSnRWKOOQHv0ZgVJjyI/O9K7dvxV2fNF/qSJ4cojvWqPiFZjMsJK4meTJBZ+UuG9+gSp4NNynE3NZShXXP2DoLH0z3k2XZvJKGwOlFOnre799JOBw7rvC5f0v+i6d7zk89Nhp6yFG3L8Um3XqSR3+DsfwXgO8gkV4ayQDcclFqtcu1bwqVfVBVOewk9RboYKsjrGSpd1SgXP+x5qc3Xex26uLhyCc/Xy4RjHCFvKw0EPubfFssNwRd3f9AmxBAiCwXT6lJHkw9MRy1hkDebTGUWr97JeR/KTx/e5iYuuJFW8BNMpRySOOw/Hm43wlJ8xI2lrCd89wAbUJx+GP0bRHsRPjlIU4PMvkJb8B+dhLm6paizLc4fhrIvKRuwhGwvOwvnZQExmrTBSoiMJ31jjc2KyqtEY3jb0eWT/u+/np8peU+ZwUf5JFN+ZAcuRI6cTK1mCywSeAGt5EQpGHZ5jZOoRZBi+nq6lIbdlf+8ruPmjjuZEnD8Cugv65fbxkt4xf9ApjU5xzYXUrf5EBzw1HXjZGvfiiWi76gBopFk9xueRHxUZ/pgHBJDj0H6mrj5msAT59fzOhWU/Ug3yX7DJ0lJbVRJX+EM/8gG09A5i3yf9CpDdKhvfYCZ1Cs6jSPqbXvFdSu+nkfDUuU04DSY5BJ6+PqQb0wQnp1+DcdqNp+uf3Fo/lhDxV2n8Tdg/WMtwbJfcnrZuqz0z9Uhd41EJ7tc8aLX2u7xe7aP9xK/2PkScPAZpSzVnaU+AQv5GZBrL9F0jNxqiR5j/6XGqgyTlspN8U+Br1PSx+PKpasI/z8j+vxnYTzWu/yTDemoZ5hf+oFtbi2k/POALOYFyBRAOWkY/5+iff/VOelpvPCjMHJH0eF2MzXQMrLizGX0OYU2cg/nCKDmnRsLvfn33X1dvCQESAw57f7j86YN6qdqUwiuW3Jpss6XAV46ku18Tkd+KMPtrQ7+jAOod8fDuEDKr+XQQA2dZatl2VPu/CeB+VSHC3w0Kz6mMMT6efgZgOHz6nJ/jGhSS8av7otSHosVAqskt1FFqmhg1q/8ROm/2hMaP1NFGZAPSVfJNbY46VPnJJMNfTUOzrTXFvxpi1azQWkJg+sU/aleQX/+LLuvNAZOcX+iNmAUeFuHvScXyPgzOzgAWUgNufvbMjPz+kUyjPzjRmXDpWe/H1PSjRtjGPp38H8fd4+5hUQAA'}
import base64
import gzip
import os
import shutil
import subprocess
import sys
import time
import uuid
from pathlib import Path

assert ACCOUNT_INDEX in (0, 1), "Use 0 for B and 1 for C"
devices = subprocess.check_output(
    ["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"], text=True).splitlines()
assert len(devices) == 2 and all("T4" in name for name in devices), "Select GPU T4 x2"
runtime = Path("/kaggle/working/qwen_runtime")
runtime.mkdir(exist_ok=True)
for name, content in PAYLOAD.items():
    (runtime / name).write_bytes(gzip.decompress(base64.b64decode(content)))
packages = ["transformers==4.57.1", "accelerate==1.10.1", "numpy==2.2.5",
            "Pillow==11.3.0", "av==16.0.1", "soundfile==0.13.1", "librosa==0.11.0"]
subprocess.run([sys.executable, "-m", "pip", "install", "--quiet", *packages], check=True)
subprocess.run([sys.executable, "-c",
    "from transformers import Qwen2_5OmniThinkerForConditionalGeneration"], check=True)
sys.path.insert(0, str(runtime))
import avh_common as avh
import avh_kaggle as kg

study, = [p.parent for p in Path("/kaggle/input").rglob("kaggle_bundle.json")
          if p.parent.joinpath("manifest.json").is_file()]
bundle = avh.read_json(study / "kaggle_bundle.json")
assert bundle["private_only"] is True
manifest_path, = Path("/kaggle/input").rglob("lean_manifest_qwen_n1000.json")
manifest = avh.read_json(manifest_path)
assert avh.digest(manifest) == "4036b97e275cca38196b05faa537bde5caa10973dda7402447000d6eaf3432c7"
assert manifest["parent_manifest_sha256"] == bundle["parent_manifest_sha256"]
index = kg.resolved_index(study)
for name, expected in bundle["files"].items():
    if not name.startswith("media/"):
        assert avh.file_hash(study / name) == expected, name
needed = {clip for row in manifest["rows"] for clip in [row["video_id"], *row["donors"]]}
avh.verify_media({clip: index[clip] for clip in needed})

from kaggle_secrets import UserSecretsClient
token = UserSecretsClient().get_secret("HF_TOKEN")
assert token, "Enable your own HF_TOKEN secret for this notebook"
launch_id = uuid.uuid4().hex
processes = []
try:
    for gpu, worker_id in enumerate(kg.account_workers(ACCOUNT_INDEX, 4)):
        out = Path(f"/kaggle/working/qwen_worker_{worker_id}")
        saved = [p.parent.parent for p in Path("/kaggle/input").rglob("predictions.jsonl")
                 if p.parent.parent.name == out.name and (p.parent / "identity.json").is_file()]
        if len(saved) > 1:
            raise RuntimeError(f"Attach one prior checkpoint for worker {worker_id}")
        if saved and not out.exists():
            shutil.copytree(saved[0] / "predictions", out / "predictions",
                            ignore=shutil.ignore_patterns("worker.lock"))
        job = dict(dataset=str(study), output=str(out), manifest=str(manifest_path),
                   cpu_verified=str(study / "cpu_verified.json"), model="qwen",
                   mode="qwen_replication", n_workers=4, account_index=ACCOUNT_INDEX,
                   worker_id=worker_id, local_gpu=gpu, launch_id=launch_id,
                   reference="fresh_unbatched_gpu", gemma_complete=True, qwen_attempt=1)
        job_path = runtime / f"job-{worker_id}.json"
        avh.atomic_json(job_path, job)
        env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(gpu),
                   TOKENIZERS_PARALLELISM="false", HF_TOKEN=token)
        with (runtime / f"worker-{worker_id}.log").open("w", encoding="utf8") as log:
            process = subprocess.Popen(
                [sys.executable, "-u", str(runtime / "avh_kaggle_worker.py"), "--job", str(job_path)],
                env=env, stdout=log, stderr=subprocess.STDOUT)
        processes.append((worker_id, process))
    while any(process.poll() is None for _, process in processes):
        time.sleep(20)
        print({worker_id: process.poll() for worker_id, process in processes}, flush=True)
finally:
    for _, process in processes:
        if process.poll() is None:
            process.terminate()
            process.wait(timeout=30)
for worker_id, process in processes:
    path = Path(f"/kaggle/working/qwen_worker_{worker_id}/completion.json")
    result = avh.read_json(path) if process.returncode == 0 and path.is_file() else {}
    print(f"QWEN_WORKER_STATUS worker={worker_id} exit={process.returncode} status={result.get('status', 'failed')}")
print("Save this private output version. Worker logs are in qwen_runtime.")
```
